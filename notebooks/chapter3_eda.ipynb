### Listing 3.4 Reporting Athena results as a pandas ```DataFrame```
__Note:__ dataset downloaded Sept 2026 contains 53,173,692 rows vs 67,435,888 in the 
book (published 2022). DC taxi dataset appears to have been updated since publication.
Data quality issues described in the book may differ from results observed here.

````python
import pandas as pd
def awscli_to_df():
    json_df = pd.read_json('/tmp/awscli.json')
    df = pd.DataFrame(json_df[0].tolist(), index = json_df.index, \
        columns = json_df[0].tolist()[0]).drop(0, axis = 0)
    
    return df
````

````python
awscli_to_df()
````
